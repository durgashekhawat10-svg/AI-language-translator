```python
from googletrans import Translator
import ipywidgets as widgets
from IPython.display import display
```

```python
translator = Translator()
```

```python
def translate_text(text,target_lang):
  translated = translator.translate(text, dest=target_lang)
  return translated.text
```

```python
text_input = widgets.Textarea(placeholder ="enter the text which you want to translate ",
                              layout=widgets.Layout(width="100%",height="100px"))
```

```python
lang_dropdown=widgets.Dropdown(
    options=[("Spanish","es"),("Hindi","hi"),("German","de"),("French","fr"),("Chinese","zh-cn")],
    descripiton="Language:"
)
```

```python
translate_button=widgets.Button(description="Translate")
output_box=widgets.Output()
```

```python
def on_translate_clicked(b):
  output_box.clear_output()
  translated_text=translate_text(text_input.value,lang_dropdown.value)
  with output_box:
    print("Translated line :",translated_text)
```

```python
translate_button.on_click(on_translate_clicked)
```

```python
display(text_input,lang_dropdown,translate_button, output_box)
```
